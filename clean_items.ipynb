**Office Items**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('officeitems.csv',skipinitialspace=True)
df = df.dropna(how='all')
print(df.head())
```

_Checking for missing values_

```python
df.isna().sum()
```

I can multiply unit cost and units to fill up any total values that are missing. 

```python
print(df[df['Total'].isna()])
```

```python
df['Total'] = df['Total'].fillna(df['Units']*df['Unit Cost'])
# print(df[df['Total'].isna()])
print(df.iloc[[16]])
```

```python
df.isna().sum()
```

Fill in the unit and the unit costs based on either average or similar entries. 

```python
df["Total"] = pd.to_numeric(df["Total"], errors="coerce")
df["Unit Cost"] = pd.to_numeric(df["Unit Cost"], errors="coerce")
df["Units"] = pd.to_numeric(df["Units"], errors="coerce")

df['Unit Cost'] = df['Unit Cost'].fillna((df['Total'])/(df['Units']))
df.isna().sum()

```

```python
print(df[df['Total'].isna()])
```

By converting all to float, I created more dirty data. Fixing this rn. 

```python
df['Total'] = df['Total'].fillna(df['Units']*df['Unit Cost'])
```

```python
print(df[df['Total'].isna()])
```

```python
df.isna().sum()
```

```python
print(df[df['Units'].isna()])
```

```python
df['Units'] = df['Units'].fillna((df['Total'])/(df['Unit Cost']))
```

```python
print(df.iloc[[13]])
```

```python
df.isna().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'],format='%m/%%d/%Y', errors='coerce')

print(df.head())
```

```python
df['Order Date'] = df['Order Date'].fillna(pd.Timestamp('2019-03-20'))

df.isna().sum()
```

For the name, I will use John Doe. 

```python
df['Name'] = df['Name'].fillna("John Doe")
print(df.head())
```

```python
df.isna().sum()
```

Data is now clean. Now to create a new csv with it. 

```python
df.to_csv("cleaned_data.csv", index=False, encoding="utf-8-sig")
```
